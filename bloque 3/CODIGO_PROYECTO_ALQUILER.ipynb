```python
datos_alquiler["ubicacion"] = datos_alquiler["ubicacion"].astype(str)
```

```python
datos_alquiler["ciudad"]   = datos_alquiler["ubicacion"].str.split(",").str[-3].str.strip()
datos_alquiler["distrito"] = datos_alquiler["ubicacion"].str.split(",").str[-4].str.strip()
datos_alquiler["barrio"]   = datos_alquiler["ubicacion"].str.split(",").str[-5].str.strip()
```

```python
datos_alquiler["distrito"] = (
    datos_alquiler["distrito"]
    .astype(str)                          # asegurar que es texto
    .str.replace("'", "", regex=False)    # eliminar comillas simples
    .str.strip()                          # quitar espacios extra
)
datos_alquiler["barrio"] = (
    datos_alquiler["barrio"]
    .astype(str)                          # asegurar que es texto
    .str.replace("'", "", regex=False)    # eliminar comillas simples
    .str.strip()                          # quitar espacios extra
)

datos_alquiler["ciudad"] = (
    datos_alquiler["ciudad"]
    .astype(str)                          # asegurar que es texto
    .str.replace("'", "", regex=False)    # eliminar comillas simples
    .str.strip()                          # quitar espacios extra
)
```

```python

datos_alquiler["distrito"] = (
    datos_alquiler["distrito"]
    .astype(str)
    .str.split(" ", n=1)   # separa en 2 partes: primera palabra + resto
    .str[1]                # se queda con la segunda parte
    .str.strip())
```

```python
datos_alquiler["barrio"] = (
    datos_alquiler["barrio"]
    .astype(str)
    .str.split(" ", n=1)   # separa en 2 partes: primera palabra + resto
    .str[1]                # se queda con la segunda parte
    .str.strip())
```
